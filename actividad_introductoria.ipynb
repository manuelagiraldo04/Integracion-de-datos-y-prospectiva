0. cargar liberías

```python
import numpy as np
import pandas as pd
```

1. conexión con la base de datos

```python
nxl = '/0. SolicitantesCrédito(Quiz 1).xlsx'
XDB = pd.read_excel(nxl , sheet_name = 0)
```

2. llamar la variable de caracterización

```python
XDe = XDB[['Edad']]
XDe = XDe.dropna()
XDe = np.array(XDe)
```

3. Se calcula la media y desviación estandar de la edad

```python
print("la media de la edad es " , np.mean(XDe))
print("la desviación estandar de la edad es " , np.std(XDe))
print("la edad minima es" , np.mean(XDe)-np.std(XDe))
print("la edad maxima es" , np.mean(XDe)+np.std(XDe))
```

4. histograma de frecuencias

```python
import matplotlib.pyplot as plt

plt.hist(XDe)
plt.show()
```

5. calcular el valor de pertenencia para una persona de 60 años

```python
VP = np.exp(-0.5*(60 - np.mean(XDe)/np.std(XDe)**2))
print("el valor de pertenencia para la edad de 60 años es: " , VP)
```
